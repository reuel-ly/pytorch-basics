### Random Tensors

Why random Tensors?

Random Tensors are important because the way neural networks learn is that they start with tensors full of random numbers and then adjust those random numbers to better understant the data.

'Start with random numbers' -> look at data -> update random numbers -> look at data -> update random numbers

```python
# Create a random tensor of size (3, 4)

random_tensor = torch.rand(3, 4)
random_tensor
```

## Matrix Multiplication

Two main ways of perofrming multiplication in neural networks and deep learning:
1. Element-wise multiplication
2. Matrix Multiplication

There are two main rules that performing matrix multiplication needs to satisfy:
1. The **inner dimension** must match:
* `(3, 2) @ (3, 2)` won't work
* `(2, 3) @ (3, 2)` will work

2. The resulting matrix has the shape of the ** outer dimensions**:
* `(2, 3) @ (3,2)` -> `(2,2)`
* `(3, 2) @ (2,3)` -> `(2, 2)`

```python
# Element wise multiplication

print(f"{tensor} * {tensor}")
print(f"equals = {tensor * tensor}")
```

```python
# Matrix Multiplication
torch.matmul(tensor, tensor)
```

### One of the most common error in deep learning

```python
# shapes for matrix multiplication
tensor_A = torch.tensor([[1,2],
                        [3,4],
                        [5, 6]])
tensor_B = torch.tensor([[7, 10],
                      [8, 11],
                      [9, 12]])
```

```python
torch.mm(tensor_A, tensor_B) # matmulll = mm
```

```python
tensor_A.shape, tensor_B.shape
```

To fix our tensor shape issues, we can manipulate the shape of one of our tensors using a **transpose**. A **transpose** switches the axes or dimensions of a given tensor.

```python
tensor_B = tensor_B.T
```

```python
torch.matmul(tensor_A, tensor_B)
```
